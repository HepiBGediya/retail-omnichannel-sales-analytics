## Business Context

This dataset represents transactional retail data across multiple years.
Each row corresponds to a product line item within an invoice.
The data includes both sales and returns, which are common in offline and online retail.

## Load Raw Data

The dataset is provided in two year-wise files and will be combined for analysis.

```python
import pandas as pd
import os

file_path = "/Users/hepigediya/Desktop/retail-omnichannel-analytics/data/raw/online_retail_II.xlsx"

# Ensure the output directory exists (use the same directory as the source Excel)
output_dir = os.path.dirname(file_path)
os.makedirs(output_dir, exist_ok=True)

df_2009 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
df_2010 = pd.read_excel(file_path, sheet_name="Year 2010-2011")

df_2009.to_csv(os.path.join(output_dir, "online_retail_2009_2010.csv"), index=False)
df_2010.to_csv(os.path.join(output_dir, "online_retail_2010_2011.csv"), index=False)
```

## Key Observations

- Dataset structure: The data is provided in two year-based sheets, which will be combined into a single transactional dataset for analysis.
- Granularity: The dataset is at line-item level, where a single invoice can contain multiple product records.
- Returns and cancellations: Negative values in the `Quantity` column represent returned items, and invoices starting with the letter 'C' indicate credit or cancellation transactions.
- Customer coverage: A significant number of records have missing `CustomerID` values, which will impact customer-level analysis and must be handled during data cleaning.
- Geography: The presence of a `Country` column enables market-level segmentation and geographic analysis.
- Revenue derivation: Total sales value is not directly available and will need to be calculated using `Quantity × UnitPrice`.
- Sales channel: The dataset does not explicitly include an offline or online sales channel indicator; this will be derived later using business rules.

Exact counts for total records, returns, missing customers, and country distribution are computed programmatically and referenced in later analysis.
